```python
# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional for better visuals
sns.set(style='whitegrid')

# Step 2: Load your data
df = pd.read_csv('staffing_data.csv')

# Step 3: Preview the data
df.head()
```

```python
# Step 4: Check data types and missing values
df.info()

# Step 5: Summary stats
df.describe()

# Step 6: See how many records you have
print(f"Number of rows: {len(df)}")
print(f"Departments: {df['Department'].unique()}")
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Department', y='Avg Time to Fill (days)', ci=None)
plt.title("Average Time to Fill by Department")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Department', y='Offer Acceptance Rate')
plt.title("Offer Acceptance Rate by Department")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
